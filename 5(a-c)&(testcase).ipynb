```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb

# Generate data
data = np.arange(1, 10, 0.01)
pdf = norm.pdf(data, loc=5.3, scale=1)

# Plot
sb.set_style('whitegrid')
sb.lineplot(x=data, y=pdf, color='black')  # ✅ specify x and y
plt.xlabel('Heights')
plt.ylabel('Probability Density')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.arange(50)

# Create y values with random noise
y = x + np.random.randint(0, 30, 50)

# Plot scatter
plt.scatter(x, y)

# Update figure parameters
plt.rcParams.update({'figure.figsize': (10, 8), 'figure.dpi': 100})

# Add labels and title
plt.title('Simple Scatter plot')
plt.xlabel('X - value')
plt.ylabel('Y - value')

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.arange(50)

# Create y values with random noise
y = x + np.random.randint(0, 30, 50)

# Update figure parameters
plt.rcParams.update({'figure.figsize': (10, 8), 'figure.dpi': 100})

# Scatter plot with color mapping
plt.scatter(x, y, c=y, cmap='Spectral')

# Add colorbar
plt.colorbar()

# Add labels and title
plt.title('Simple Scatter plot')
plt.xlabel('X - value')
plt.ylabel('Y - value')

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Define data
y = pd.Series([1, 2, 3, 4, 3, 5, 4])
x = pd.Series([1, 2, 3, 4, 5, 6, 7])

# Calculate correlation
correlation = y.corr(x)
print(correlation)

# Scatter plot
plt.scatter(x, y)

# Fit and plot best-fit line
plt.plot(np.unique(x), np.poly1d(np.polyfit(x, y, 1))(np.unique(x)), color='red')

# Show plot
plt.show()
```

```python
from numpy.random import randn, seed
from scipy.stats import pearsonr

# Set random seed for reproducibility
seed(1)

# Generate two datasets
data1 = 20 * randn(1000) + 100
data2 = data1 + (10 * randn(1000) + 50)

# Calculate Pearson correlation
corr, _ = pearsonr(data1, data2)

# Print result
print('Pearson correlation:', corr)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Generate data
np.random.seed(1)
study_hours = np.random.normal(5, 1.5, 100)
scores = 10 * study_hours + np.random.randn(100) * 5

# Correlation
print("Correlation matrix:\n", np.corrcoef(study_hours, scores))
print("Pearson correlation:", pearsonr(study_hours, scores)[0])

# Plot
plt.scatter(study_hours, scores)
m, b = np.polyfit(study_hours, scores, 1)
plt.plot(study_hours, m*study_hours + b)
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.title("Study Hours vs Exam Scores")
plt.show()
```
